#### From observation some date cols have been stored as object types. We need to correct this before exporting to BigQuery

```python
# Convert date cols to date dtypes (date only)
date_cols = [
    "Policy_Start_Date",
    "Policy_End_Date",
    "Payment_Date",
    "Premium_Date",
    "Claim_Date",
    "Payout_Date",
    "Application_Date",
    "Submitted_Date",
    "Reviewed_Date"
]

for key, value in tables.items():
  for col in date_cols:
    if col in value.columns:
      # Convert to datetime first, then extract the date component
      value[col] = pd.to_datetime(value[col]).dt.date
```

```python
from datetime import date
# Export data to BigQuery
import os
from google.cloud import bigquery

# Set up credentials
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = r"/content/drive/MyDrive/Datasets/alx-data-community-498121-464a202ca657.json"


def load_dfs_to_BigQuery(tables, project_id, dataset_id):
    client = bigquery.Client(project=project_id)

    # Configure the job to truncate (overwrite) the table
    job_config = bigquery.LoadJobConfig(
        write_disposition=bigquery.WriteDisposition.WRITE_TRUNCATE
    )

    for table_name, df in tables.items():
        table_ref = f"{project_id}.{dataset_id}.{table_name}"

        job = client.load_table_from_dataframe(df, table_ref, job_config=job_config)
        job.result()  # Wait for the load to finish



```

```python
# Export the data into BigQuery

load_dfs_to_BigQuery(tables=tables,
                     project_id='alx-data-community-498121',
                     dataset_id='insurance_dataset')
```
